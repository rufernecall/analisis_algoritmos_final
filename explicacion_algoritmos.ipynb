### 1. Ordenamiento y Búsqueda (Semanas 1-2)
Probamos el **Quick Sort** para el inventario y el **Merge Sort** para los rankings de ventas.

```python
def quick_sort_estudiante(lista):
    if len(lista) <= 1:
        return lista
    # Elegimos el pivote del centro como vimos en la diapositiva
    pivote = lista[len(lista) // 2]
    menores = [x for x in lista if x < pivote]
    iguales = [x for x in lista if x == pivote]
    mayores = [x for x in lista if x > pivote]
    return quick_sort_estudiante(menores) + iguales + quick_sort_estudiante(mayores)

datos = [101, 120, 105, 110, 115]
print("Inventario ordenado por ID:", quick_sort_estudiante(datos))
```

### 2. Optimización con Programación Dinámica (Semana 5)
Validación del Algoritmo de la **Mochila 0/1** para la reposición de stock con presupuesto limitado.

```python
def mochila_dp(capacidad, pesos, valores):
    n = len(valores)
    # Creamos la tabla de memorización (Tabulación)
    tabla = [[0 for _ in range(capacidad + 1)] for _ in range(n + 1)]
    
    for i in range(1, n + 1):
        for w in range(1, capacidad + 1):
            if pesos[i-1] <= w:
                # Comparamos si conviene llevar el producto o no
                tabla[i][w] = max(valores[i-1] + tabla[i-1][w - pesos[i-1]], tabla[i-1][w])
            else:
                tabla[i][w] = tabla[i-1][w]
    
    return tabla[n][capacidad]

precios_compra = [10, 20, 30] # Costos de los productos
prioridades = [6, 10, 8]     # Nivel de importancia
presupuesto = 40

print("Maxima prioridad obtenida:", mochila_dp(presupuesto, precios_compra, prioridades))
```

### 3. Simulación de Riesgo (Montecarlo - Semana 6)
En lugar de calcular Pi, simulamos la probabilidad de que el stock llegue a cero.

```python
import random

def simular_riesgo(stock, promedio_venta, dias):
    exitos = 0
    simulaciones = 1000
    
    for _ in range(simulaciones):
        stock_temp = stock
        for _ in range(dias):
            # Simulación de venta diaria aleatoria
            stock_temp -= random.randint(0, promedio_venta * 2)
            if stock_temp <= 0:
                exitos += 1
                break
    
    return (exitos / simulaciones) * 100

print(f"Riesgo de agotamiento en 7 dias: {simular_riesgo(20, 5, 7):.2f}%")
```

### 4. Rutas de Entrega (Grafos BFS - Semana 7)
Buscamos la ruta más corta (mínimo de saltos) entre el almacén y una tienda.

```python
from collections import deque

def ruta_corta(grafo, inicio, fin):
    visitados = {inicio: [inicio]}
    cola = deque([inicio])
    
    while cola:
        nodo = cola.popleft()
        if nodo == fin:
            return visitados[nodo]
        
        for vecino in grafo.get(nodo, []):
            if vecino not in visitados:
                visitados[vecino] = visitados[nodo] + [vecino]
                cola.append(vecino)
    return None

red_locales = {
    'Almacen': ['Tienda_1', 'Tienda_2'],
    'Tienda_1': ['Tienda_3'],
    'Tienda_2': ['Tienda_3'],
    'Tienda_3': []
}

print("Mejor ruta para el delivery:", ruta_corta(red_locales, 'Almacen', 'Tienda_3'))
```
