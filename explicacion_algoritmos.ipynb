```python
# --- 1. Algoritmos de Ordenamiento ---

def quick_sort(lista):
    if len(lista) <= 1:
        return lista
    pivote = lista[len(lista) // 2]
    izq = [x for x in lista if x < pivote]
    centro = [x for x in lista if x == pivote]
    der = [x for x in lista if x > pivote]
    return quick_sort(izq) + centro + quick_sort(der)

datos = [38, 27, 43, 3, 9, 82, 10]
print("Lista ordenada con Quick Sort:", quick_sort(datos))
```

```python
# --- 2. Busqueda Binaria ---

def busqueda_binaria(lista, objetivo):
    izq, der = 0, len(lista) - 1
    while izq <= der:
        medio = (izq + der) // 2
        if lista[medio] == objetivo:
            return medio
        elif lista[medio] < objetivo:
            izq = medio + 1
        else:
            der = medio - 1
    return -1

lista_ordenada = [2, 3, 4, 10, 40]
print("El numero 10 esta en la posicion:", busqueda_binaria(lista_ordenada, 10))
```

```python
# --- 3. Programacion Dinamica (Mochila) ---

def mochila_01(capacidad, pesos, valores, n):
    K = [[0 for _ in range(capacidad + 1)] for _ in range(n + 1)]
    for i in range(n + 1):
        for w in range(capacidad + 1):
            if i == 0 or w == 0:
                K[i][w] = 0
            elif pesos[i-1] <= w:
                K[i][w] = max(valores[i-1] + K[i-1][w - pesos[i-1]], K[i-1][w])
            else:
                K[i][w] = K[i-1][w]
    return K[n][capacidad]

v = [60, 100, 120]
p = [10, 20, 30]
print("Valor maximo en la mochila:", mochila_01(50, p, v, 3))
```

```python
# --- 4. Backtracking (8 Reinas) ---

def es_seguro(tablero, fila, col):
    for i in range(col):
        if tablero[fila][i] == 1: return False
    for i, j in zip(range(fila, -1, -1), range(col, -1, -1)):
        if tablero[i][j] == 1: return False
    for i, j in zip(range(fila, len(tablero)), range(col, -1, -1)):
        if tablero[i][j] == 1: return False
    return True

def resolver(tablero, col):
    if col >= len(tablero): return True
    for i in range(len(tablero)):
        if es_seguro(tablero, i, col):
            tablero[i][col] = 1
            if resolver(tablero, col + 1): return True
            tablero[i][col] = 0
    return False

t = [[0]*8 for _ in range(8)]
if resolver(t, 0):
    print("Solucion encontrada para las 8 reinas.")
    for f in t: print(f)
```
